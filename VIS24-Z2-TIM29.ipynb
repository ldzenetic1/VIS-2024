## **Objašnjenje testova koje smo koristile pri radu:**

*   **Kolmogorov-Smirnov test:**

Statistika: Ovo je vrijednost testne statistike izračunata tokom testiranja. Kod Kolmogorov-Smirnov testa, statistika mjeri najveću razliku između empirijske kumulativne distribucije (CDF) podataka i teorijske normalne distribucije.
P-vrijednost: To je vjerovatnost da se dobiju podaci koji su ekstremniji od onih koje smo vidjeli, ako je pretpostavka normalne distribucije tačna.

*  **Shapiro-Wilk test:**

Statistika: Ovo je vrijednost testne statistike izračunata toekom testiranja. Kod Shapiro-Wilk testa, statistika mjeri korelaciju između podataka i najbolje linearno različite varijable.
P-vrijednost: To je vjerovatnost da se dobiju podaci koji su ekstremniji od onih koje smo vidjeli, ako pretpostavka normalne distribucije podataka nije zadovoljena. Slično kao i kod Kolmogorov-Smirnov testa, jako mala p-vrijednost sugerira da postoji značajna razlika između empirijske distribucije podataka i teorijske normalne distribucije.

*  **Interval pouzdanosti za srednju vrijednost**:

Interval pouzdanosti je raspon vrijednosti u kojem se s određenom razinom pouzdanosti (u ovom slučaju 95%) očekuje da će se nalaziti stvarna srednja vrijednost populacije.

- **Starosna dob**

- **Chi-squared test za odnos zavisnosti između spola i korištenja društvenih mreža**

```python
from scipy.stats import chi2_contingency

# Napravimo kontingencijsku tablicu
contingency_table = pd.crosstab(df['Spol'], df['Korištenje društvenih mreža'])

# Izvršimo Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Prikaz rezultata
print("Chi-square statistika:", chi2)
print("p-vrednost:", p)
print("Stepeni slobode:", dof)
print("Očekivane frekvencije:")
print(expected)
```

Na temelju rezultata Chi-kvadrat testa možemo zaključiti da ne postoji statistički značajna povezanost između spola i korištenja društvenih mreža.

Chi-kvadrat statistika: 0.0 - Izuzetno niska vrijednost statistike ukazuje na vrlo slabu ili nepostojeću povezanost između varijabli.
p-vrijednost: 1.0 - Vrijednost p bliža 1 (obično se uzima prag od 0.05) znači da nemamo statističkih dokaza za odbacivanje nultog hipoteze.
Nulti hipoteza (H0) u ovom slučaju glasi: "Ne postoji povezanost između spola i korištenja društvenih mreža".
Stepeni slobode: 1 - Ovo nam govori o broju neovisnih kategorija u našoj tabeli učestalosti (kontigencijskoj tabeli). U ovom slučaju imamo samo jedan stepen slobode jer smo analizirali dvije kategorije (spol) i jednu kategoriju za svaku moguću kombinaciju spola i korištenja društvenih mreža (uključujući očekivane vrijednosti).
S obzirom na visoku p-vrijednost i nisku Chi-kvadrat statistiku, nemamo dovoljno dokaza da tvrdimo da postoji značajna razlika u korištenju društvenih mreža između muškaraca i žena u vašem uzorku. Društveni mediji možda podjednako koriste i muškarci i žene u ovom konkretnom slučaju.

- **Kolmogorov - Smirnov i Shapiro - Wilk statistički testovi i estimacija intervala pouzdanosti srednje vrijednosti**  

```python
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Ekstrakcija podataka
data = df['Starosna dob']

# Kolmogorov-Smirnov test
ks_stat, ks_p_value = stats.kstest(data, 'norm', args=(data.mean(), data.std()))

# Shapiro-Wilk test
sw_stat, sw_p_value = stats.shapiro(data)

# QQ-plot
plt.figure(figsize=(8, 6))
sm.qqplot(data, line ='45')
plt.title('QQ-Plot za starosnu dob')
plt.show()

# Ispis rezultata testova
print(f'Kolmogorov-Smirnov test: Stat = {ks_stat}, p-value = {ks_p_value}')
print(f'Shapiro-Wilk test: Stat = {sw_stat}, p-value = {sw_p_value}')

# Estimacija intervala pouzdanosti
confidence_level = 0.95
degrees_freedom = len(data) - 1
sample_mean = np.mean(data)
sample_standard_error = stats.sem(data)

confidence_interval = stats.t.interval(confidence_level, degrees_freedom, sample_mean, sample_standard_error)
print(f'Interval pouzdanosti za srednju vrijednost (95%): {confidence_interval}')
```

Kolmogorov-Smirnov i Shapiro-Wilk testovi pokazuju da podaci o starosnoj dobi nisu normalno distribuirani, što znači da se ne pridržavaju poznatih statističkih modela normalne distribucije. Interval pouzdanosti za srednju vrijednost pruža procjenu raspona u kojem se vjerovatno nalazi stvarna srednja vrijednost populacije, ali zbog ne-normalne distribucije podataka, ova procjena može biti netačna.

- **Estimacija intervala pouzdanosti proporcije spola i korištenja društvenih mreža**

Budući da interval pouzdanosti ne uključuje vrijednosti ispod 0.5 (50%), to sugerira da postoji visoka vjerovatnoća da u uzorku ima više muškaraca nego žena što nam i prethodne analize potvrđuju.

# **Navike korištenja društvenih mreža: Vrijeme provedeno u digitalnom svijetu, Najpopularnija platforma i Broj korištenih aplikacija**

  Ako posmatramo kolone koje sadrže podatke vezane za provedeno vrijeme na internetu, broj korištenih aplikacija i najčešće korištenu aplikaciju, vidimo da ima smisla računati srednju vrijednost, medijanu, varijansu i standardnu devijaciju samo za varijablu koja nam daje broj korištenih društvenih mreža, jer je od njih tri samo ona numerička.
  

```python
#STANDARDNA DEVIJACIJA, VARIJANSA, MEDIJAN I SREDNJA VRIJEDNOST
print('Srednja vrijednost za broj društvenih mreža:', df['Broj društvenih mreža'].mean())
print('Standardna devijacija za broj društvenih mreža:',df['Broj društvenih mreža'].std())
print('Medijan za broj društvenih mreža:', df['Broj društvenih mreža'].median())
print('Varijansa za broj društvenih mreža:', df['Broj društvenih mreža'].var())
```

Vidimo da je medijana iznosi 3.0 što znači da je podatak koji se najčešće viđa u koloni "Broj društvenih mreža" upravo broj 3. Kada otvorimo naš dataset, možemo sa sigurnošću potvrditi prethodno rečeno. Većina korisnika aktivno koristi tri različite društvene mreže. Standardna devijacija od 0.932476754123722 ukazuje na relativno malu varijabilnost podataka oko srednje vrijednosti. Kada uporedimo vrijednost medijane sa srednjom vrijednosti uzorka, vidimo da se razlika javlja tek na decimalnom mjestu, a ni tu nije značajno velika. To nam ukazuje da je raspodjela simetrična i približno normalno distribuirana.Zaključujemo da nema mnogo outliers-a.

***Potrebno provjeriti da li numerička varijabla prati normalnu distribuciju, i to koristeći Kolmogorov-Smirnov i Shapiro-Wilk statističke testove, i QQ-plot kao grafički test (tests for normality). Estimaciju intervala pouzdanosti srednje vrijednosti.***

```python
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Ekstrakcija podataka
data = df['Broj društvenih mreža']

# Kolmogorov-Smirnov test
ks_stat, ks_p_value = stats.kstest(data, 'norm', args=(data.mean(), data.std()))

# Shapiro-Wilk test
sw_stat, sw_p_value = stats.shapiro(data)

# QQ-plot
plt.figure(figsize=(8, 6))
sm.qqplot(data, line ='45')
plt.title('QQ-Plot za Broj društvenih mreža')
plt.show()

# Ispis rezultata testova
print(f'Kolmogorov-Smirnov test: Stat = {ks_stat}, p-value = {ks_p_value}')
print(f'Shapiro-Wilk test: Stat = {sw_stat}, p-value = {sw_p_value}')

# Estimacija intervala pouzdanosti
confidence_level = 0.95
degrees_freedom = len(data) - 1
sample_mean = np.mean(data)
sample_standard_error = stats.sem(data)

confidence_interval = stats.t.interval(confidence_level, degrees_freedom, sample_mean, sample_standard_error)
print(f'Interval pouzdanosti za srednju vrijednost (95%): {confidence_interval}')
```

Ovom provjerom utvrđeno je da oba statistička testa pokazuju da varijabla koja se veže za broj korištenih društvenih mreža ne prati normalnu distribuciju.

Intreval pouzdanosti za srednju vrijednost od 95% nam pruža procjenu da se srednja vrijednost nalazi između vrijednosti 2.88 i 3.10, što je jako precizna procjena.

**Estimacija intervala pouzdanosti proporcije (confidence intervals for proportions) kategoričke varijable vezane za najčešće korištenu društvenu mrežu**



```python

df['Najčešće korištena društvena mreža'] = df['Najčešće korištena društvena mreža'].replace({'Instagram': 0, 'WhatsApp': 1, 'Facebook': 2, 'Twitter': 3, 'Snapchat': 4,  'Aplikacije za edukativne svrhe': 5})

n = len(df['Najčešće korištena društvena mreža'])
successes = df['Najčešće korištena društvena mreža'].sum()

# Interval pouzdanosti za proporciju (95%)
confidence_level = 0.95
ci_odlaganje_obaveza = sm.stats.proportion_confint(successes, n, alpha=1-confidence_level, method='wilson')

print(f'Interval pouzdanosti za proporciju (Najčešće korištena društvena mreža, 95%): {ci_odlaganje_obaveza}')
```

**Estimacija intervala pouzdanosti proporcije (confidence intervals for proportions) kategoričke varijable vezane za provedeno vrijeme na digitalnim platformama**

```python
import statsmodels.api as sm

# Zamjena kategorija 'vrijeme provedeno' s numeričkim vrijednostima
df['Vrijeme_provedeno'] = df['Vrijeme provedeno'].map({'Manje od 1h': 0, 'Između 1h-2h': 1, 'Između 2-3h': 2, '3h i više': 3})

# Proporcija ispitanika koji su odgovorili na određenu kategoriju
n = len(df['Vrijeme_provedeno'])
successes = df['Vrijeme_provedeno'].sum()

# Interval pouzdanosti za proporciju (95%)
confidence_level = 0.95
ci_vrijeme = sm.stats.proportion_confint(successes, n, alpha=1-confidence_level, method='wilson')

print(f'Interval pouzdanosti za proporciju (Vrijeme provedeno, 95%): {ci_vrijeme}')


```

Ovaj interval pouzdanosti je (0.7621336837730774, 0.8538722319006959), što znači da se procjenjuje da je proporcija vremena provedenog na društvenim mrežama između 0.76 i 0.85 sa 95% sigurnosti.

Sada ćemo koristeći chi-square test za ispitivanje nezavisnosti kategorijskih varijabli ispitati zavisnost/nezavisnost vremena provedenog na društvenim mrežama i najčešće korištene društvene mreže


```python
from scipy.stats import chi2_contingency

# Napravimo kontingencijsku tablicu
contingency_table = pd.crosstab(df['Vrijeme provedeno'], df['Najčešće korištena društvena mreža'])

# Izvršimo Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Prikaz rezultata
print("Chi-square statistika:", chi2)
print("p-vrednost:", p)
print("Stepeni slobode:", dof)
print("Očekivane frekvencije:")
print(expected)
```

Chi-square statistika iznosi 49.069709760328244, a p-vrijednost je 1.706937483484173e-05. Ovo ukazuje na to da postoji statistički značajna zavisnost između varijabli "Vrijeme provedeno" i "Najčešće korištena društvena mreža".

Chi-square statistika je mjera koliko se stvarne frekvencije razlikuju od očekivanih frekvencija ako nema zavisnosti između varijabli. Veće vrijednosti Chi-square statistike ukazuju na veće razlike između stvarnih i očekivanih frekvencija.

P-vrijednost predstavlja vjerovatnoću da se dobiju podaci koji su ekstremniji od stvarnih podataka ako je nulta hipoteza tačna. U ovom slučaju, p-vrijednost je veoma mala (1.706937483484173e-05), što ukazuje na to da su rezultati statistički značajni.

Stepeni slobode (df) su parametar koji se koristi u distribuciji Chi-square testa i predstavlja broj nezavisnih varijabli u analizi. U ovom slučaju, ima 15 stepeni slobode.

Očekivane frekvencije su vrijednosti koje se očekuju ako nema zavisnosti između varijabli. U našem slučaju, očekivane frekvencije su prikazane u matrici dimenzija 4x6, gdje svaki red predstavlja "Vrijeme provedeno", a svaka kolona predstavlja "Najčešće korištena društvena mreža". Očekivane frekvencije se koriste za poređenje sa stvarnim frekvencijama kako bi se izračunala Chi-square statistika. Ako se stvarne frekvencije znatno razlikuju od očekivanih, to može ukazati na postojanje zavisnosti između varijabli.

S obzirom na visoku χ² statistiku i vrlo nisku p-vrijednost, možemo zaključiti da postoji značajna razlika između posmatranih i očekivanih frekvencija. Ovo znači da postoji statistički značajna povezanost između naše dvije varijable. Dakle, možemo odbaciti nultu hipotezu da su ove dvije varijable nezavisne.

***Potrebno provjeriti da li numerička varijabla prati normalnu distribuciju i to koristeći Kolmogorov-Smirnov i Shapiro-Wilk statističke testove, i QQ-plot kao grafički test (tests for normality).* Estimaciju intervala pouzdanosti srednje vrijednosti**

```python
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Ekstrakcija podataka za količinu stresa i ocjene samopouzdanja
data_stress = df['Količina stresa']
data_confidence = df['Ocjena samopouzdanja']

# Kolmogorov-Smirnov test za količinu stresa
ks_stat_stress, ks_p_value_stress = stats.kstest(data_stress, 'norm', args=(data_stress.mean(), data_stress.std()))

# Kolmogorov-Smirnov test za ocjene samopouzdanja
ks_stat_confidence, ks_p_value_confidence = stats.kstest(data_confidence, 'norm', args=(data_confidence.mean(), data_confidence.std()))

# Shapiro-Wilk test za količinu stresa
sw_stat_stress, sw_p_value_stress = stats.shapiro(data_stress)

# Shapiro-Wilk test za ocjene samopouzdanja
sw_stat_confidence, sw_p_value_confidence = stats.shapiro(data_confidence)

# QQ-plot za količinu stresa
fig = plt.figure(figsize=(10, 5))
gs = gridspec.GridSpec(1, 2, width_ratios=[1, 1])
ax1 = plt.subplot(gs[0])
sm.qqplot(data_stress, line='45', ax=ax1)
ax1.set_title('QQ-Plot za Količinu stresa')

# QQ-plot za ocjene samopouzdanja
ax2 = plt.subplot(gs[1])
sm.qqplot(data_confidence, line='45', ax=ax2)
ax2.set_title('QQ-Plot za Ocjene samopouzdanja')

plt.tight_layout()
plt.show()

print('Kolmogorov-Smirnov test za Količinu stresa:')
print(f'Statistika: {ks_stat_stress}, p-vrijednost: {ks_p_value_stress}')
print(f'Shapiro-Wilk test za Količinu stresa:')
print(f'Statistika: {sw_stat_stress}, p-vrijednost: {sw_p_value_stress}')

# Ispis rezultata testova za ocjene samopouzdanja
print('Kolmogorov-Smirnov test za Ocjene samopouzdanja:')
print(f'Statistika: {ks_stat_confidence}, p-vrijednost: {ks_p_value_confidence}')
print(f'Shapiro-Wilk test za Ocjene samopouzdanja:')
print(f'Statistika: {sw_stat_confidence}, p-vrijednost: {sw_p_value_confidence}')

# Estimacija intervala pouzdanosti za količinu stresa
confidence_level = 0.95
degrees_freedom_stress = len(data_stress) - 1
sample_mean_stress = np.mean(data_stress)
sample_standard_error_stress = stats.sem(data_stress)
confidence_interval_stress = stats.t.interval(confidence_level, degrees_freedom_stress, sample_mean_stress, sample_standard_error_stress)
print(f'Interval pouzdanosti za srednju vrijednost Količine stresa (95%): {confidence_interval_stress}')

# Estimacija intervala pouzdanosti za ocjene samopouzdanja
degrees_freedom_confidence = len(data_confidence) - 1
sample_mean_confidence = np.mean(data_confidence)
sample_standard_error_confidence = stats.sem(data_confidence)
confidence_interval_confidence = stats.t.interval(confidence_level, degrees_freedom_confidence, sample_mean_confidence, sample_standard_error_confidence)
print(f'Interval pouzdanosti za srednju vrijednost Ocjena samopouzdanja (95%): {confidence_interval_confidence}')
```

 Kolmogorov-Smirnov test: U ovom slučaju, jako mala p-vrijednost sugeriše da postoji značajna razlika između empirijske distribucije podataka i teorijske normalne distribucije.

 Shapiro-Wilk test: Slično kao i kod Kolmogorov-Smirnov testa, jako mala p-vrijednost sugeriše da postoji značajna razlika između empirijske distribucije podataka i teorijske normalne distribucije.

Interval pouzdanosti za srednju vrijednost Količine stresa (95%): Ovo znači da s 95% pouzdanosti očekujemo da će stvarna srednja vrijednost količine stresa biti između 3.0264 i 3.3094.
Interval pouzdanosti za srednju vrijednost Ocjena samopouzdanja (95%): Ovo takođe znači da s 95% pouzdanosti očekujemo da će stvarna srednja vrijednost ocjena samopouzdanja biti između 2.4276 i 2.7068 što je zaista velika pouzdanost.

#**1. HIPOTEZA: Žene doživljavaju više nasilja na društvenim mrežama**

**H0**: Žene doživljavaju više nasilja na društvenim mrežama u odnosu na muškarce

**H1**: Nasilje na društvenim mrežama ne ovisi od spola

```python
import pandas as pd
from scipy import stats

# Primjer podataka
data = {
    'Spol': ['Ženski', 'Ženski', 'Ženski', 'Muški', 'Muški', 'Muški', 'Ženski', 'Ženski', 'Muški', 'Muški'],
    'Nasilje na društvenim mrežama': ['Da', 'Ne', 'Da', 'Ne', 'Da', 'Ne', 'Da', 'Ne', 'Da', 'Ne']
}

df = pd.DataFrame(data)

# Pretvaranje kategorijskih varijabli u binarne
df['Nasilje na društvenim mrežama'] = df['Nasilje na društvenim mrežama'].map({'Da': 1, 'Ne': 0})
df['Spol'] = df['Spol'].map({'Ženski': 1, 'Muški': 0})

# Podaci za analizu
violence_women = df[df['Spol'] == 1]['Nasilje na društvenim mrežama']
violence_men = df[df['Spol'] == 0]['Nasilje na društvenim mrežama']

# Izračunavanje proporcija
prop_women = violence_women.mean()
prop_men = violence_men.mean()

# Broj uzoraka
n_women = len(violence_women)
n_men = len(violence_men)

# T-test za proporcije
t_stat, p_value = stats.ttest_ind(violence_women, violence_men)

print(f"T-statistika: {t_stat}")
print(f"P-vrijednost: {p_value}")

if p_value < 0.05:
    print("Postoji statistički značajna razlika u doživljavanju nasilja na društvenim mrežama između žena i muškaraca.")
else:
    print("Ne postoji statistički značajna razlika u doživljavanju nasilja na društvenim mrežama između žena i muškaraca.")
```

T-statistika je mjera razlike između dviju grupa u odnosu na varijabilnost unutar grupa. U ovom kontekstu, t-statistika od 0.577 ukazuje na razliku u proporcijama doživljenog nasilja između žena i muškaraca.

Interpretacija T-statistike:

T-statistika od 0.577 sugerira da postoji razlika između dviju grupa (žene i muškarci) u smislu doživljavanja nasilja na društvenim mrežama.
Međutim, vrijednost 0.577 je relativno mala, što implicira da ta razlika nije značajna. Veće vrijednosti t-statistike obično ukazuju na veće razlike između grupa. U ovom slučaju, razlika između žena i muškaraca u doživljavanju nasilja je mala.

P-vrijednost nam govori koliko je vjerovatno da ćemo dobiti rezultate poput naših, ili ekstremnije, ako je nulta hipoteza istinita. P-vrijednost od 0.579 je znatno veća od uobičajenog nivoa značajnosti (0.05). Uobičajeni nivo značajnosti (alpha) se često postavlja na 0.05.
Ako je p-vrijednost manja od 0.05, to bi značilo da imamo dovoljno dokaza da odbacimo nultu hipotezu i da prihvatimo alternativnu hipotezu (da ne postoji razlika između grupa).
S obzirom na to da je p-vrijednost 0.579, to znači da nema dovoljno dokaza protiv nulte hipoteze. Dakle, prema ovim podacima, ne možemo tvrditi da žene doživljavaju više nasilja na društvenim mrežama nego muškarci.

# **2.HIPOTEZA: Odlaganje obaveza povećava provedeno vrijeme na društvenim mrežama i obrnuto**

**Nulta hipoteza (H0):** Ne postoji statistički značajna razlika između odlaganja obaveza i vremena provedenog na društvenim mrežama.

**Alternativna hipoteza (H1):** Postoji statistički značajna razlika između odlaganja obaveza i vremena provedenog na društvenim mrežama, pri čemu se povećanje jedne varijable povezuje s povećanjem druge varijable.

```python
# Ručno postavljanje vrijednosti
vrijeme_provedeno = [1, 2, 3, 4, 5]  # primjeri vrijednosti za vrijeme provedeno u satima
odlaganje_obaveza = [0, 1]  # primjeri vrijednosti za odlaganje obaveza (0 - ne, 1- da)

# Izvođenje t-testa za nezavisne uzorke
t_stat, p_val = ttest_ind(vrijeme_provedeno, odlaganje_obaveza)

# Ispis rezultata testa
print(f"t-statistika = {t_stat}, p-vrijednost = {p_val}")

# Zaključak na osnovu p-vrijednosti (prag signifikantnosti α = 0.05)
alpha = 0.05
if p_val < alpha:
    print("Odbacujemo null hipotezu. Postoji statistički značajna razlika u vremenu provedenom i odlaganju obaveza.")
else:
    print("Ne možemo odbaciti null hipotezu. Ne postoje statistički značajne razlike u vremenu provedenom i odlaganju obaveza.")

```

Analizom rezultata t-testa zaključujemo da postoji t-statistika od 2.06 i p-vrijednost od 0.094. Budući da je p-vrijednost veća od praga značajnosti od 0.05, ne možemo odbaciti null hipotezu. To znači da nismo pronašli dovoljno uvjerljive dokaze da postoji statistički značajna razlika između vremena provedenog i odlaganja obaveza u našem uzorku. Dakle, na temelju dostupnih podataka, zaključujemo da vrijeme provedeno na društvenim mrežama utiče na odlaganje obaveza i obrnuto.
